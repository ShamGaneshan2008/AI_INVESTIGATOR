```python
evidence = """
02:14 AM - Alex logged into the company server.
02:15 AM - Login came from an unfamiliar IP address.
02:18 AM - Financial database was accessed.
02:21 AM - 2500 financial records were downloaded.
02:25 AM - Alex's laptop was found switched off.
02:30 AM - Alex said he was sleeping at home.
"""

with open("evidence.txt", "w") as file:
    file.write(evidence)
```

```python
events_data = []

# Re-initialize 'events' from 'evidence' to ensure it's the expected list of strings
events = evidence.strip().split("\n")

for event_str in events:
    if not event_str: # Skip empty lines
        continue
    parts = event_str.split("-")
    if len(parts) < 2:
        print(f"Warning: Skipping malformed event line: {event_str}")
        continue

    time = parts[0].strip()
    message = parts[1].strip()

    events_data.append({
        "time": time,
        "message": message
    })

# Display the structured evidence
for event_dict in events_data:
    print(f"{event_dict['time']} - {event_dict['message']}")
```

```python
for event in events_data:
    if "database" in event["message"]:
        print(event)
```

```python
suspicious_events = []

keywords = ["database", "downloaded", "unfamiliar"]

for event in events_data:
    for keyword in keywords:
        if keyword in event["message"].lower():
            suspicious_events.append(event)
            break

print("Suspicious events:")

for event in suspicious_events:
    print(event)


# Create timeline
timeline = sorted(suspicious_events, key=lambda event: event["time"])


# Compare consecutive events
for i in range(len(timeline) - 1):
    current = timeline[i]
    next_event = timeline[i + 1]

    print("CURRENT:", current["message"])
    print("NEXT:", next_event["message"])
    print()
```

```python
for i in range(len(timeline) - 1):
  current = timeline[i]
  next_event = timeline[i+1]

  if "database" in next_event["message"].lower() and "downloaded" in current["message"].lower():
    print(current)
    print(next_event)
    print()
```

```python
for i in range(len(timeline) - 2):
  first = timeline[i]
  second = timeline[i + 1]
  third = timeline[i + 2]

  if("unfimilar" in first["message"].lower()
    and "database" in second["message"].lower()
    and "database" in third["message"].lower()):

    print("Suspicious chain Detected")
    print(first["time"], "->", first["message"])
    print(second["time"], "->", second["message"])
    print(third["time"], "->", third["message"])
```

```python
for event in events_data:
  message = event["message"].lower()

  if "sleeping at home" in message:
    print("Statement Found: ")
    print(event)
```

```python
login_event = None
statement_event = None

for event in events_data:
  message = event["message"].lower()

  if "login" in message:
    login_event = event
  elif "sleeping at home" in message:
    statement_event = event

if login_event and statement_event:
  print("Possible Suspicous activity")
  print("Login:", login_event["time"], "->", login_event["message"])
  print("Statement", statement_event["time"], "->", statement_event["message"])
```

```python
for event_item in events_data:
  score = 0
  message = event_item["message"].lower()

  if "unfamilier" in message:
    score += 2

  if "database" in message:
    score += 2

  if "dowloaded" in message:
    score += 3

  event_item["score"] = score


for event in events_data:
  print(event["time"], "->", event["message"], "->", event["score"])
```
